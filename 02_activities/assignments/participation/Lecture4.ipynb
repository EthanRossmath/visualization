```python
# load libraries

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import PIL
import requests

# sample library
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75, 50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 100, 240, 99, 220])
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2,
                               nrows=1,
                               figsize = (7,3)
                               )
ax1.scatter(x1, y1)
ax2.bar(x2, y2)
```

```python
# Activity

fig, (ax1, ax2) = plt.subplots(ncols=2,
                               nrows=1,
                               figsize = (7,3)
                               )
ax1.scatter(x1, y1, color = 'pink')
ax2.bar(x2, y2, color = 'indigo')

ax1.set_title('A homogeneous image of some dots')
ax2.set_title('# of people named after \n One Piece characters')
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                    ['ax2', 'ax3']],
                                    figsize = (7,4)
                                   )
someaxes['ax1'].scatter(x1, y1)
someaxes['ax2'].bar(x2, y2)
someaxes['ax3'].plot(x1, y1)
```

```python
# sample data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 100, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

# define our figure and axes
fig, ax = plt.subplots(figsize = (7, 3))

# now call both bar and plot elements to the same axes
ax.bar(x, y1,
       color = 'indigo')
ax.plot(x, y2,
        color = 'red')
```

```python
y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize = (7, 3))
ax.plot(x, y2, color='red')
ax.errorbar(x, # x data
            y2, # y data
            yerr=y2_sd, # vertical error bars
            fmt='none', # not sure
            ecolor='indigo', # color
            elinewidth=4, # width of line
            capsize=6, # horizontal width of the caps
            capthick=2, # thickness of caps
            errorevery=2 # skips n - 1
             )
```

```python
from PIL import Image
from io import BytesIO
import requests
```

```python
response = requests.get('https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png')
image_file = BytesIO(response.content)
image = Image.open(image_file)
```

```python
fig, ax = plt.subplots(figsize = (7, 3))
ax.plot(x, y2, color = 'red')

ax_image = fig.add_axes([0.1, # x coordinate (ON FIGURE, NOT AXES)
                         0.11, # y coordinate (ON FIGURE, NOT AXES)
                         0.15, # image width
                         0.35 # image height
                         ])

ax_image.imshow(image)
ax_image.axis('off')
```
